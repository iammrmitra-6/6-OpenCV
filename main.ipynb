```python
# converting it into grayscale image
import cv2

img = cv2.imread("Mango.png")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite("Mango_gray.png", img_gray)
print("Saved grayscale image as Mango_gray.png")


print("Original image shape:", img.shape)
print("Grayscale image shape:", img_gray.shape)
```

```python
#Playing with RGB color channels--
import cv2

img = cv2.imread("Mango.png")

#img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

img[:,:,2]=0

imgblue=img[:,:,0]
imggreen=img[:,:,1]
imgred=img[:,:,2]

cv2.imwrite("Mango_blue.png", imgblue)
cv2.imwrite("Mango_green.png", imggreen)
cv2.imwrite("Mango_red.png", imgred)

print("Original image shape:", img.shape)
print("Grayscale image shape:", img_gray.shape)
```

```python
# Playing with RGB color channels
import cv2

img = cv2.imread("Mango.png")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Keep only the red channel
img_red = img.copy()
img_red[:, :, 0] = 0
img_red[:, :, 1] = 0

cv2.imwrite("Mango_red_channel.png", img_red)

print("Original image shape:", img.shape)
print("Grayscale image shape:", img_gray.shape)
print("Saved red-channel image as Mango_red_channel.png")
```

```python
# Resize an image
import cv2

img = cv2.imread("Mango.png")

img_resize = cv2.resize(img, (256, 256))
cv2.imwrite("Mango_resize.png", img_resize)

print("Original shape:", img.shape)

print("Resized shape:", img_resize.shape)
print("Saved resized image as Mango_resize.png")
```

```python
# Show the original dimension and the resized dimension clearly
import cv2

img = cv2.imread("Mango.png")
if img is None:
    raise FileNotFoundError("Mango.png not found in the current folder.")

print("Original is:", img.shape)
img_half = cv2.resize(img, (img.shape[1] // 10, img.shape[0] // 10))
print("After resize is:", img_half.shape)
cv2.imwrite("Mango_half_resize.png", img_half)
print("Saved half-size file: Mango_half_resize.png")
```

```python
#6.flipping
#1-horizontal , 0-verical , -1-both
import cv2

img = cv2.imread("Mango.png")

img_flip=cv2.flip(img,0)
cv2.imwrite("Mango_flip.png", img_flip)


print("Saved fliped image as Mango_resize.png")
```

```python
#cropping
import cv2

img = cv2.imread("Mango.png")

img_crop=img[0:2000,0:2000]#basic sclising
cv2.imwrite("img_crop.png", img_crop)
```

```python
# 10. Drawing shapes (rectangles, circles, lines) and text
import cv2
import numpy as np

img = np.zeros((512, 512, 3), dtype=np.uint8)

cv2.rectangle(img, (100, 100), (300, 300), (250, 0, 0), -1)
cv2.circle(img, (100, 400), 50, (0, 0, 255), -1)
cv2.line(img, (0, 0), (512, 512), (0, 255, 0), 2)
cv2.putText(img, "hello", (100, 100), cv2.FONT_ITALIC, 2, (0, 255, 255), 2, cv2.LINE_AA)

cv2.imwrite("drawn_shapes.png", img)
print("Saved shape drawing image as drawn_shapes.png")
```
